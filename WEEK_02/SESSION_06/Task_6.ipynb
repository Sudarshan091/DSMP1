### **`Problem 3:`** Write a Python function that accepts a string and calculate the number of upper case letters and lower case letters.

```
Sample String : 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'
Expected Output :
No. of Upper case characters :  9
No. of Lower case Characters :  47
```

````python
# Write code here
def problem3_count_case(string):
    """Count uppercase and lowercase letters."""
    upper = sum(1 for c in string if c.isupper())
    lower = sum(1 for c in string if c.islower())
    return upper, lower
print("PROBLEM 3: Count Case Letters")
sample = 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'
upper, lower = problem3_count_case(sample)
print(f"String: {sample}")
print(f"Uppercase: {upper}, Lowercase: {lower}")
print()
````

### **`Problem 5:`** Write a Python function to check whether a number is perfect or not.

A Perfect number is a number that is half the sum of all of its positive divisors (including itself).

Example :

```
The first perfect number is 6, because 1, 2, and 3 are its proper positive divisors, and 1 + 2 + 3 = 6.
Equivalently, the number 6 is equal to half the sum of all its positive divisors: ( 1 + 2 + 3 + 6 ) / 2 = 6.

The next perfect number is 28 = 1 + 2 + 4 + 7 + 14. This is followed by the perfect numbers 496 and 8128.
```


````python
# Write code here
def problem5_is_perfect(n):
    """Check if a number is perfect."""
    if n <= 0:
        return False
    divisor_sum = sum(i for i in range(1, n) if n % i == 0)
    return divisor_sum == n

# Test
print("PROBLEM 5: Check Perfect Numbers")
test_nums = [6, 28, 496, 8128, 10]
for num in test_nums:
    print(f"{num}: {problem5_is_perfect(num)}")
print()
````

### **`Problem-6:`** Write a Python function to concatenate any no of dictionaries to create a new one.

```
Sample Dictionary :
dic1={1:10, 2:20}
dic2={3:30, 4:40}
dic3={5:50,6:60}
Expected Result : {1: 10, 2: 20, 3: 30, 4: 40, 5: 50, 6: 60}
```

````python
# Write code here
def problem6_merge_dicts(*dicts):
    """Merge multiple dictionaries."""
    result = {}
    for d in dicts:
        result.update(d)
    return result

# Test
print("PROBLEM 6: Merge Dictionaries")
d1 = {1:10, 2:20}
d2 = {3:30, 4:40}
d3 = {5:50, 6:60}
merged = problem6_merge_dicts(d1, d2, d3)
print(f"Dict1: {d1}, Dict2: {d2}, Dict3: {d3}")
print(f"Merged: {merged}")
print()
````

`Problem-7` Write a python function that accepts a string as input and returns the word with most occurence.

```
Input:
hello how are you i am fine thank you
```

```
Output
you -> 2
```

````python
# Write code here
def problem7_most_frequent_word(string):
    """Find the most frequently occurring word."""
    from collections import Counter
    words = string.split()
    word_counts = Counter(words)
    max_count = max(word_counts.values())
    return [(w, max_count) for w, c in word_counts.items() if c == max_count]

# Test
print("PROBLEM 7: Most Frequent Word")
input7 = "hello how are you i am fine thank you"
result = problem7_most_frequent_word(input7)
print(f"Input: {input7}")
for word, count in result:
    print(f"Output: {word} -> {count}")
print()
````

`Problem-8` Write a python function that receives a list of integers and prints out a histogram of bin size 10

```
Input:
[13,42,15,37,22,39,41,50]
```

```
Output:
{11-20:2,21-30:1,31-40:2,41-50:3}
```

````python
# Write code here

def problem8_histogram(numbers, bin_size=10):
    """Create histogram with bin size."""
    histogram = {}
    for num in numbers:
        bin_start = (num // bin_size) * bin_size
        bin_end = bin_start + (bin_size - 1)
        bin_key = f"{bin_start+1}-{bin_end}"
        histogram[bin_key] = histogram.get(bin_key, 0) + 1
    return dict(sorted(histogram.items()))

# Test
print("PROBLEM 8: Histogram with Bin Size 10")
input8 = [13, 42, 15, 37, 22, 39, 41, 50]
print(f"Input: {input8}")
print(f"Output: {problem8_histogram(input8)}")
print()



````

`Problem-9` Write a python function that accepts a list of 2D co-ordinates and a query point, and then finds the the co-ordinate which is closest in terms of distance from the query point.

```
List of Coordinates
[(1,1),(2,2),(3,3),(4,4)]
Query Point
(0,0)
```

```
Output
Nearest to (0,0) is (1,1)
```

````python
# Write code here
def problem9_nearest_coordinate(coords, query):
    """Find nearest coordinate using Euclidean distance."""
    import math
    min_dist = float('inf')
    nearest = None
    for coord in coords:
        dist = math.sqrt((coord[0] - query[0])**2 + (coord[1] - query[1])**2)
        if dist < min_dist:
            min_dist = dist
            nearest = coord
    return nearest, min_dist

# Test
print("PROBLEM 9: Find Nearest Coordinate")
coords = [(1,1), (2,2), (3,3), (4,4)]
query = (0, 0)
nearest, dist = problem9_nearest_coordinate(coords, query)
print(f"Coordinates: {coords}, Query: {query}")
print(f"Output: Nearest is {nearest} with distance {dist:.2f}")
print()
````

`Problem-10`:Write a python program that receives a list of strings and performs bag of word operation on those strings

https://en.wikipedia.org/wiki/Bag-of-words_model

````python
# Write code here
def problem10_bag_of_words(strings):
    """Create bag of words model."""
    from collections import Counter
    all_words = []
    for string in strings:
        words = string.lower().split()
        all_words.extend(words)
    return dict(Counter(all_words))

# Test
print("PROBLEM 10: Bag of Words")
strings = ["the cat sat on the mat", "the dog sat on the log"]
bow = problem10_bag_of_words(strings)
print(f"Strings: {strings}")
print(f"Bag of Words: {bow}")
print()

````

###`Problem-12:`Write a Python program to create a list containing the power of said number in bases raised to the corresponding number in the index using Python map.
`Input:`
```
list1 = [1,2,3,4,5,6]
```
`Output:`
```
[1,2,9,64,625,-]
```


````python
# Write code here

def problem12_power_list(numbers):
    """Raise each number to power of its index."""
    return list(map(lambda x: x[1] ** x[0], enumerate(numbers)))

# Test
print("PROBLEM 12: Power List (element^index)")
list12 = [1, 2, 3, 4, 5, 6]
output12 = problem12_power_list(list12)
print(f"Input: {list12}")
print(f"Output: {output12}")
print()
````

###`Problem-13` Using filter() and list() functions and .lower() method filter all the vowels in a given string.



````python
# Write code here

def problem13_filter_vowels(string):
    """Remove all vowels from string."""
    vowels = 'aeiou'
    return ''.join(filter(lambda x: x.lower() not in vowels, string))

# Test
print("PROBLEM 13: Filter Vowels")
string13 = "Hello World! This is Amazing."
result13 = problem13_filter_vowels(string13)
print(f"Input: {string13}")
print(f"Output (vowels removed): {result13}")
print()
````

`Problem-14`: Use reduce to convert a 2D list to 1D

````python
# Write code here

def problem14_flatten_2d(list_2d):
    """Convert 2D list to 1D using reduce."""
    from functools import reduce
    return reduce(lambda x, y: x + y, list_2d)

# Test
print("PROBLEM 14: Flatten 2D List using reduce()")
list2d = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
output14 = problem14_flatten_2d(list2d)
print(f"Input: {list2d}")
print(f"Output: {output14}")
print()
````
