# Session 1: Python Basics - Short Notes


## 1. About Python
- High-level, interpreted programming language.
- Easy to read and write.
- Widely used in AI, ML, web development, automation.


## 2. print() Function
- Used to display output on the screen.
- Syntax: `print("Hello")`
- Useful for debugging.


## 3. Data Types
- int → Whole numbers (e.g., 5)
- float → Decimal numbers (e.g., 3.14)
- str → Text (e.g., "Hello")
- bool → True or False
- Use `type()` to check data type.


## 4. Variables
- Used to store data.
- Syntax: `x = 10`
- Variable name should not start with a number.
- Case-sensitive (age ≠ Age).


## 5. Comments
- Used to explain code.
- Single-line: `# This is a comment`
- Multi-line: Triple quotes `''' comment '''`


## 6. Keywords and Identifiers
- Keywords: Reserved words (e.g., if, else, while, def).
- Identifiers: Names given to variables/functions.
- Cannot use keywords as identifiers.


## 7. User Input
- `input()` is used to take input from user.
- Always returns data as string.
- Example: `name = input("Enter name: ")`


## 8. Type Conversion
- Convert one data type to another.
- `int()`, `float()`, `str()`
- Example: `age = int(input("Enter age: "))`


## 9. Literals
- Fixed values written directly in code.
- Examples: `10`, `3.14`, "Hello", `True`
